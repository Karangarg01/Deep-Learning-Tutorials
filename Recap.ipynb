```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data: y = 3x + 2 + noise
np.random.seed(42)
X = np.random.rand(100, 1)
y = 3 * X + 2 + 0.1 * np.random.randn(100, 1)

# Initialize weights
W = np.random.randn(1)
b = np.random.randn(1)

lr = 0.1   # learning rate
epochs = 200
loss_history = []

for epoch in range(epochs):
    # Predictions
    y_pred = W * X + b

    # Loss (Mean Squared Error)
    loss = np.mean((y_pred - y) ** 2)
    loss_history.append(loss)

    # Gradients
    dW = np.mean(2 * (y_pred - y) * X)
    db = np.mean(2 * (y_pred - y))

    # Update weights
    W -= lr * dW
    b -= lr * db

print("Trained W:", W)
print("Trained b:", b)

# Plot loss
plt.plot(loss_history)
plt.title("Loss vs Epochs")
plt.show()
```

This cell imports `numpy` for numerical operations and `matplotlib.pyplot` for plotting. It then generates synthetic data that follows a linear relationship (`y = 3x + 2 + noise`) to demonstrate a simple linear regression problem.

Here, the model's initial weights (`W`) and bias (`b`) are randomly initialized. Hyperparameters like the learning rate (`lr`) and the number of training epochs are also defined. A list `loss_history` is created to store the loss at each epoch.

This `for` loop implements the gradient descent algorithm to train the linear regression model. In each epoch, it calculates predictions, computes the Mean Squared Error (MSE) loss, calculates the gradients with respect to `W` and `b`, and then updates the weights and bias using the learning rate.

After training, the final trained values for `W` (weight) and `b` (bias) are printed. These values represent the learned parameters of the linear model.

Finally, this code plots the `loss_history` over the epochs. This plot helps visualize how the model's loss decreased during training, indicating whether the training process was successful.
